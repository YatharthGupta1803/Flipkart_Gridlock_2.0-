# **Exploratory Data Analysis for the provided training Data.**

**Key Observations**
1. RoadTypes are not complete, almost 324 rows are incomplete.
2. TimeStamp is not correctly formatted
3. Large Vehicle is only in "Allowed" or "Not Allowed" can be further converted into binary.
4. Landmarks is given only in "Yes" or "No" can be further converted into binary.
5. Missing Temperature at few Rows
6. Missing Weather Category at few rows.
7. Need to Validate the Data_Types.

```python
import pandas as pd
import numpy as np
```

Processing --


```python
test = pd.read_csv('test.csv')
print(test)
```

**A .** Dropping the Empty Rows with no value for 'Road Type' column

```python
test.dropna(subset=['RoadType'], inplace=True)

print(test['RoadType'])
```

**B.** Fixing the Empty Temperature value by taking a grouped average

```python
test['Temperature'] = test.groupby('timestamp')['Temperature'].transform(lambda x: x.fillna(x.mean()))
```

**C.** Fixing Weather as If it's Rainy at 2:00 PM in a specific geohash, it's likely Rainy at 2:15 PM.

```python
test['Weather'] = test.groupby('day')['Weather'].ffill().bfill()
```

**D.** Converting the Large Vehicles column to Binary

```python
test['LargeVehicles'] = test['LargeVehicles'].transform(lambda x: 1 if x=="Allowed" else 0)
```

**E.** Converting the Landmarks to Binary

```python
test['Landmarks'] = test['Landmarks'].transform(lambda x: 1 if x=="Yes" else 0)
```

**F.** Formatting the Time Stamp into 2 different columns hour and Minute.

```python
test['Hour'] = test['timestamp'].apply(
    lambda x: int(x.split(':')[0])
)
test['Minute'] = test['timestamp'].apply(
    lambda x: int(x.split(':')[1])
)
```

**G.** Saving the final Data in a new file.

```python
test.to_csv('test_cleaned.csv', index=False)
```
